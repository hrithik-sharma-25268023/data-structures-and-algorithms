## Counting the digits of a number

```python
def count_all_digits(number: int) -> int:
    count = 0
    n = number
    while n>0:
        count+=1
        n=n//10
    return count
```

```python
count_all_digits(5873)
```

## Reversing and checking the number is palindrome

```python
def check_palindrome(number: int) -> int:
    """checks palindrome"""

    n = number
    result = 0
    while n > 0:
        last_digit = n%10
        result = result*10 + last_digit
        n = n//10
    return result
```

## Armstrong Number

```python
def is_armstrong_number(number: int) -> bool:
    """checks if number is armstrong"""

    num = number
    count = 0
    total_sum = 0
    while num > 0:
        count+=1
        num = num//10

    num = number
    while num > 0:
        digit = num%10
        total_sum += digit**count
        num=num//10

    return True if total_sum == number else False
```

```python
is_armstrong_number(1634)
```
