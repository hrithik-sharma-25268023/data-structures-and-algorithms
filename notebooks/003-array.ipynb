## Array Implimentaion

```python
from typing import List

class Array:

    def __init__(self, size: int) -> None:
        """main array method"""

        self.array = [None] * size
        self.size = size


    def insertion_into_array(self, length: int) -> List:
        """return the array after insertion"""

        i = 0
        while i <= length-1:
            ele = int(input(f"Enter element at index {i}:- "))
            self.array[i] = ele
            i = i+1
        return self.array


    @property
    def length_of_array(self) -> int:
        """length of array"""

        counter = 0
        for ele in self.array:
            if ele != None:
                counter +=1
        return counter


    def traverse_array(self) -> None:
        """travarse the array"""

        if self.length_of_array == 0:
            return "Please insert some elements into array"
        else:
            for i in range(self.length_of_array):
                print(f"Element at index {i}:- {self.array[i]}")


    def insert_element(self, index: int, element: int) -> List:
        """insert a element at given index"""

        self.array = self.array+[None]
        i = self.length_of_array - 1
        while i >= index:
            self.array[i+1] = self.array[i]
            i = i-1
        self.array[index] = element
        return self.array


    def delete_element(self, index: int):
        """delete a element at given index"""

        self.array[index] = None
        for i in range(index, self.size - 1):
            self.array[i] = self.array[i+1]
        return self.array[:self.size-1]
            
```

```python
def sum_of_elements(array: List) -> int:
    """find sum of all elements in an array"""

    sum_ = 0
    for i in range(len(array)):
        sum_ = sum_ + array[i]
    return sum_
```

```python
def maximum_element(array: List) -> int:
    """maximum element from an array"""

    maximum = array[0]
    for i in range(len(array)):
        if array[i] > maximum:
            maximum = array[i]
    return maximum
```

## Linear Search

```python
def linear_search(size: int, key: int) -> str | None:
    """impliments linear search"""

    array = [None]*size
    i = 0
    while i<size:
        ele = int(input(f"Enter element at index {i}:- "))
        array[i] = ele
        i = i+1

    for i in range(size):
        if key == array[i]:
            result = i
            return result
    return -1
```

```python
def run_linear_search(size: int, key: int) -> None:

    result = linear_search(size=size, key=key)
    if result == -1:
        print("Search unsuccessful, element not found")
    else:
        print(f"Search successful, element found at index {result}.")
```

```python
run_linear_search(size=5, key=2)
```
