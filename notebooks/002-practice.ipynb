## Count the Digits

```python
def count_the_digits(number: int) -> dict:
    """method to count the digits"""

    count = 0
    n = number
    digits = []
    sum_of_digits = 0
    product_of_digits=1
    while n>0:
        count+=1
        last_digit = n%10
        sum_of_digits+=last_digit
        product_of_digits*=last_digit
        digits.append(last_digit)
        n = n//10
    return {'count':count, 'digits':digits, 'sum':sum_of_digits, 'product':product_of_digits}
```

```python
count_the_digits(532)
```

## Reverse and Palindrome

```python
def get_reverse_and_check_palindrome(number: int) -> dict:
    """get reverse and palindrome"""

    reverse = 0
    n = number
    while n>0:
        last_digit = n%10
        reverse = reverse*10 + last_digit
        n = n//10
    return {'reverse': reverse, 'palindrome':reverse==number}
```

```python
get_reverse_and_check_palindrome(532)
```

```python
get_reverse_and_check_palindrome(1221)
```

## Occurence of Digit

```python
def count_occurence(number: int) -> None:
    """counts the occurence of each digit in a number"""

    n = number
    digits = []
    frequency = {}
    while n>0:
        last_digit = n%10
        digits.append(last_digit)
        n = n//10

    for num in digits:
        if num not in frequency:
            frequency[num] = 1
        else:
            frequency[num] += 1
    return frequency
```

```python
count_occurence(122111)
```

## Table

```python
def table(number: int) -> None:
    """table"""

    for i in range(1,11):
        print(f"{number} X {i} = {number*i}")
```

```python
table(15)
```

## Sum of n natural numbers

```python
def sum_n(n: int) -> int:
    """sum of n natural numbers"""

    total = 0
    for i in range(1, n+1):
        total = total + i

    return total
```

```python
sum_n(5)
```

```python
def fact(n: int) -> int:
    """factorial"""

    factorial = 1
    for i in range(1, n+1):
        factorial = factorial*i

    return factorial
```
