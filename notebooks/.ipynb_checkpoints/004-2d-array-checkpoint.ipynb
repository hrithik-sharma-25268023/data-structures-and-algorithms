## 2D Array

```python
from typing import List, Tuple

class Array2D:

    def __init__(self, n_rows: int, n_cols: int) -> None:
        """makes 2D arrays"""

        self.n_rows = n_rows
        self.n_cols = n_cols
        self.array = [[None]*self.n_cols for row in range(self.n_rows)]


    @property
    def shape(self) -> Tuple:
        """return shape of 2D array"""

        return (self.n_rows, self.n_cols)


    def insertion_into_array(self) -> List:
        """insertion into 2D array"""

        for i in range(self.n_rows):
            for j in range(self.n_cols):
                element = int(input(f"Enter element at position {i}, {j} :-"))
                self.array[i][j] = element
        return self.array
    
```

## 2D Array Operations 

```python
arr_1 = Array2D(n_rows=3, n_cols=3)
arr_2 = Array2D(n_rows=3, n_cols=3)
```

```python
def add_arrays(array_1: Array2D, array_2: Array2D) -> List:
    """add arrays"""

    if array_1.shape == array_2.shape:
        arr_1 = array_1.insertion_into_array()
        arr_2 = array_2.insertion_into_array()
        for i in range(array_1.shape[0]):
            for j in range(array_2.shape[1]):
                arr_2[i][j] = arr_2[i][j] + arr_1[i][j]
        return arr_2
    else:
        print(f"Size Mismatch Error.")
```

```python
add_arrays(arr_1, arr_2)
```

```python
def transpose(array: Array2D) -> List:
    """transpose of a matrix"""

    new = [[None]*array.n_rows for row in range(array.n_cols)]
    for i in range(array.shape[1]):
        for j in range(array.shape[0]):
            new[i][j] = array.array[j][i]
    return new
```

```python
arr_t = Array2D(2,3)
arr_transpose = arr_t.insertion_into_array()
```
